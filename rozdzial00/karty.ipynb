```python
import random
suits = ["trefl","pik","kler","karo"]
faces = ["walet","dama","krol","as"]
numbers = [2,3,4,5,6,7,8,9,10]
```

```python
suits
```

```python
def draw():
    the_suit = random.choice(suits)
    the_card = random.choice(faces + numbers)
    return the_card, " ", the_suit
```

```python
draw()
```
